```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
file_path = '/content/drive/MyDrive/ Heart Disease Prediction/heart_disease_data.csv'
heart_data = pd.read_csv(file_path)
print(heart_data.head())  # Display first 5 rows to confirm it loaded correctly
```

```python
print(heart_data.shape)       # Number of rows and columns
print(heart_data.info())      # Data types and missing values info
print(heart_data.isnull().sum())  # Check for missing values
print(heart_data.describe())  # Statistical summary
print(heart_data['target'].value_counts())  # Target variable distribution
```

```python
X = heart_data.drop(columns='target', axis=1)  # Features
Y = heart_data['target']                        # Target label
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y, test_size=0.2, stratify=Y, random_state=2)
```

```python
model = LogisticRegression(max_iter=1000)
model.fit(X_train, Y_train)
```

```python
X_train_pred = model.predict(X_train)
training_accuracy = accuracy_score(Y_train, X_train_pred)
print(f'Accuracy on Training Data: {training_accuracy:.4f}')
```

```python
X_test_pred = model.predict(X_test)
test_accuracy = accuracy_score(Y_test, X_test_pred)
print(f'Accuracy on Test Data: {test_accuracy:.4f}')
```

```python
input_data = (62, 0, 0, 140, 268, 0, 0, 160, 0, 3.6, 0, 2, 2)

input_data_np = np.asarray(input_data).reshape(1, -1)
prediction = model.predict(input_data_np)

if prediction[0] == 0:
    print('The Person does NOT have Heart Disease.')
else:
    print('The Person has Heart Disease.')
```
